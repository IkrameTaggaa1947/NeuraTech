```python
import os
import pandas as pd

BASE_FOLDER = r"C:\AgorAI Hackathon\كلمات"

dfs_parts = {}

for word_index, word in enumerate(os.listdir(BASE_FOLDER), start=1):
    word_path = os.path.join(BASE_FOLDER, word)
    if not os.path.isdir(word_path):
        continue

    for part_index, file in enumerate(os.listdir(word_path), start=1):
        if not file.endswith(".csv"):
            continue

        # Format DataFrame name as dfX_partY
        var_name = f"df{word_index}_part{part_index}"

        file_path = os.path.join(word_path, file)
        df = pd.read_csv(file_path, skiprows=1)

        # Dynamically create variables with the desired format
        globals()[var_name] = df
        dfs_parts[var_name] = df
```

```python
import numpy as np

# Initialize lists to store EEG data and labels
X = []  # 3D array for EEG data
y = []  # Labels

# Iterate through all DataFrames to extract EEG data and labels
for var_name in list(globals().keys()):
    if var_name.startswith("df") and "_part" in var_name:
        df = globals()[var_name]
        if isinstance(df, pd.DataFrame):
            # Ensure required columns are present
            required_columns = ['EEG.Counter', 'EEG.AF3', 'EEG.F7', 'EEG.F3', 'EEG.FC5', 'EEG.T7', 'EEG.P7',
                                'EEG.O1', 'EEG.O2', 'EEG.P8', 'EEG.T8', 'EEG.FC6', 'EEG.F4', 'EEG.F8', 'EEG.AF4']
            if all(col in df.columns for col in required_columns):
                # Group data into 1-second chunks (128 samples per chunk)
                num_chunks = len(df) // 128
                for i in range(num_chunks):
                    # Exclude EEG.Counter
                    chunk = df.iloc[i * 128:(i + 1) *
                                    128][required_columns[1:]].to_numpy()
                    X.append(chunk)
                    # Assign label based on the DataFrame name (e.g., df1_part1 → label 1)
                    # Extract word number from dfX_partY
                    label = int(var_name.split("_part")[0][2:])
                    y.append(label)

# Convert lists to numpy arrays
# Shape: (num_chunks_total, num_channels, num_samples_per_chunk)
X = np.array(X)
y = np.array(y)  # Shape: (num_chunks_total,)

print(f"EEG data shape: {X.shape}")
print(f"Labels shape: {y.shape}")
```

```python
# Find columns where the quality is 0 or 1 and return corresponding details
results = []

for var_name in list(globals().keys()):
    if var_name.startswith("df") and "_part" in var_name:
        df = globals()[var_name]
        if isinstance(df, pd.DataFrame):
            for column in df.columns:
                if column.startswith("EQ."):
                    quality_issues = df[df[column].isin([0, 1])]
                    if not quality_issues.empty:
                        for _, row in quality_issues.iterrows():
                            results.append({
                                "Channel": column,
                                # Extract word number
                                "Word": var_name.split("_part")[0][2:],
                                # Extract participant number
                                "Participant": var_name.split("_part")[1],
                                "Row Index": row.name,
                                "Quality": row[column]
                            })

# Display the results
for result in results:
    print(f"Channel: {result['Channel']}, Word: {result['Word']}, Participant: {result['Participant']}, Row Index: {result['Row Index']}, Quality: {result['Quality']}")
```

```python
# Drop rows where the quality is 0 or 1 for EQ columns and return the new shape of X
filtered_X = []
filtered_y = []

for var_name in list(globals().keys()):
    if var_name.startswith("df") and "_part" in var_name:
        df = globals()[var_name]
        if isinstance(df, pd.DataFrame):
            # Ensure required EQ columns are present
            eq_columns = [col for col in df.columns if col.startswith("EQ.")]
            if eq_columns:
                # Filter rows where all EQ columns have quality > 1
                valid_rows = df[~df[eq_columns].isin([0, 1]).any(axis=1)]
                if not valid_rows.empty:
                    # Group data into 1-second chunks (128 samples per chunk)
                    num_chunks = len(valid_rows) // 128
                    for i in range(num_chunks):
                        chunk = valid_rows.iloc[i *
                                                128:(i + 1) * 128][eq_columns].to_numpy()
                        filtered_X.append(chunk)
                        # Assign label based on the DataFrame name (e.g., df1_part1 → label 1)
                        label = int(var_name.split("_part")[0][2:])
                        filtered_y.append(label)

# Convert lists to numpy arrays
# Shape: (num_chunks_total, num_channels, num_samples_per_chunk)
filtered_X = np.array(filtered_X)
filtered_y = np.array(filtered_y)  # Shape: (num_chunks_total,)

print(f"Filtered EEG data shape: {filtered_X.shape}")
```
